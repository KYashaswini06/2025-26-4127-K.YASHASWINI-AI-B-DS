```python
#Dictionaries
import pandas as pd
data={
"ID":[101,102,103,104,105],
"Name":["Alice","Bob","Charlie","David","Eva"],
"Age":[22,25,23,24,26],
"Marks":[85,78,92,88,90]
}
df=pd.DataFrame(data)
print(df)
print(df["Name"])
print(df["Marks"])
df[["Name","Marks"]]
print("Location 0:",df.iloc[0])
print("Location 1 t0 4:",df.iloc[1:4])
print("Location 1 to 3:",df.iloc[0:3,1:3])
```
