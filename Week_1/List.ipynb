```python
#List
lst=[1,2,3,2]
lst.append(4)
print("After append:",lst)
lst.extend([5,6])
print("After extend:",lst)
lst.insert(2, 10)
print("After insert:",lst)
lst.remove(2)
print("After remove:",lst)
lst.pop()
print("Poped value:", lst)
print("Index of 10:", lst.index(10))
print("Count of 2:", lst.count(2))
lst.sort()
print("Sorted list:",lst)
lst.reverse()
print("Reversed list:",lst)
print("Maximum value:",max(lst))
print("Minimum value:",min(lst))
```
