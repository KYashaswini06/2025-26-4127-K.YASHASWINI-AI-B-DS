```python
#Tuple
t1=(10,20,30,40,50)
t2=(60,70,80)
t3=("Apple","Banana","Cherry")
print("Tuple t1:",t1)
print("Tuple t2:",t2)
print("Tuple t3:",t3)
print("First element of t1:",t1[0])
print("Last element of t1:",t1[-1])
print("Slice t1[1:4]:",t1[1:4])
t4=t1+t2
print("Concatenated tuple t4:",t4)
t5=t3*2
print("Repeated tuple t5:",t5)
print("Length of t1:",len(t1))
print("Count of 20 in t1:",t1.count(20))
print("Element in t1:")
for item in t1:
  print(item ,end=" ")
print(t1[2])
print(t2[2])
print(t1[-1])
print(t3[-2])
#Tuple unpacking
tup = (1,2,3)
a,b,c=tup
print(a)
print(b)
print(c)
print("Membership:\n", 30 in t1)
print(100 in t1)
```
