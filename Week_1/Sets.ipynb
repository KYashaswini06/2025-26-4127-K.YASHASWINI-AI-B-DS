```python
#Set operations
s1={1,2,3,4,5,6}
s2={2,3,4,5,6}
print("Union of s1 and s2:",s1.union(s2))
print("Intersection of s1 and s2:",s1.intersection(s2))
print("Difference of s1 and s2 is:",s1.difference(s2))
print("Difference of s2 and s1 is:",s2.difference(s1))
print("Symmetric difference of s1 and s2 is:",s1.symmetric_difference(s2))
print("The subset of:",s1.issubset(s2))
print("Disjoint set:",s1.isdisjoint(s2))
print("The superset of:",s1.issuperset(s2))
```
